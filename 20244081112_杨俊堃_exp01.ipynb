```python
# 1. 四个运算符：求值 + 观察类型
results = [
    ("8 / 3",  8 / 3,   "真除法，结果一定是 float"),
    ("8 // 3", 8 // 3,  "向下取整除法，结果是 int"),
    ("8 % 3",  8 % 3,   "取模（求余数），结果是 int"),
    ("2 ** 7", 2 ** 7,  "幂运算，结果是 int"),
]

for expr, value, meaning in results:
    print(f"{expr:<8} = {value!r:<8} type = {type(value).__name__:<5} # {meaning}")
```

```python
# 2. 四种基本类型
i = 42
f = 3.14
s = "Python"
b = True

for value in (i, f, s, b):
    print(f"{value!r:<10} -> {type(value)}")
```

```python
# 3. 类型转换
print('int("123")   ->', repr(int("123")), type(int("123")))
print('float("3.14")->', repr(float("3.14")), type(float("3.14")))
print("int(3.9)     ->", repr(int(3.9)), type(int(3.9)), "  # 向零截断，不是四舍五入")

try:
    int("goat")
except ValueError as e:
    print()
    print("捕获到异常:", type(e).__name__)
    print("异常信息:", e)
```

**转换结果解释**

- `int("123")` → `123`：字符串内容是合法的十进制整数，可以转换。
- `float("3.14")` → `3.14`：字符串内容是合法的浮点数，可以转换。
- `int(3.9)` → `3`：浮点转整数是**向零截断**（truncate），直接丢掉小数部分，不是四舍五入。
- `int("goat")` 抛出 `ValueError: invalid literal for int() with base 10: 'goat'`：
  因为字符串 `"goat"` 不能解释成十进制整数，`int()` 无法从文本中得到数值。
  注意这里抛的是 `ValueError`（值不合法）而不是 `TypeError`（类型不合法），
  因为 `"goat"` 的类型是 `str`，恰好是 `int()` 接受的类型，只是**内容**不合法。

## 实验2：函数、条件、循环与递归

**思路**：把每道小题封装成函数，用边界值（100/90/80/60/0 以及越界值）测试条件分支；
循环用 for 和 while 两种写法互相验证；递归与迭代互相验证。

```python
def score_level(score):
    """返回成绩等级；超出 0~100 范围时抛出 ValueError。"""
    if isinstance(score, bool) or not isinstance(score, (int, float)):
        raise ValueError(f"成绩必须是数值，收到：{score!r}")
    if score < 0 or score > 100:
        raise ValueError(f"成绩超出 0~100 范围：{score}")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"


print("--- 正常范围 ---")
for s in [100, 95, 90, 89.9, 80, 79, 60, 59.9, 0]:
    print(f"{s:>6} -> {score_level(s)}")

print()
print("--- 越界范围 ---")
for bad in [-1, 101, 200]:
    try:
        score_level(bad)
    except ValueError as e:
        print(f"{bad:>6} -> 捕获 ValueError: {e}")
```

```python
# 2. for 与 while 两种方式求 1~100 之和
total_for = 0
for i in range(1, 101):
    total_for += i

total_while = 0
i = 1
while i <= 100:
    total_while += i
    i += 1

print("for 循环结果  :", total_for)
print("while 循环结果:", total_while)
print("两者一致      :", total_for == total_while)
print("内置 sum 校验 :", sum(range(1, 101)))
print("公式 n(n+1)/2 :", 100 * 101 // 2)
```

```python
# 3. 阶乘：递归版与迭代版
def factorial_recursive(n):
    """递归求 n!，约定 0! = 1。"""
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)


def factorial_iterative(n):
    """迭代求 n!，约定 0! = 1。"""
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result


print(f"{'n':>3} {'递归':>12} {'迭代':>12}  一致")
for n in [0, 1, 5, 10]:
    r = factorial_recursive(n)
    it = factorial_iterative(n)
    print(f"{n:>3} {r:>12} {it:>12}  {r == it}")

import math
print()
print("与 math.factorial 对比:", all(factorial_recursive(n) == math.factorial(n) for n in range(0, 15)))
```

```python
# 4. summary(values)：以元组返回最小值、最大值、平均值
def summary(values):
    """接收非空数值列表，返回 (最小值, 最大值, 平均值) 元组。"""
    if not values:
        raise ValueError("values 不能为空")
    return min(values), max(values), sum(values) / len(values)


sample = [85, 58, 92, 76, 60, 89]
lo, hi, avg = summary(sample)          # 元组解包
print("样本   :", sample)
print("最小值 :", lo)
print("最大值 :", hi)
print(f"平均值 : {avg:.4f}")

whole = summary(sample)
print("不解包直接看返回类型:", type(whole).__name__, "->", whole)

try:
    summary([])
except ValueError as e:
    print("空列表 -> 捕获 ValueError:", e)
```

```python
# 5. sum_all(*args)：返回所有参数之和
def sum_all(*args):
    """返回所有参数之和；无参数时返回 0。"""
    return sum(args)


print("sum_all()                  =", sum_all())
print("sum_all(10)                =", sum_all(10))
print("sum_all(1, 2, 3, 4, 5)     =", sum_all(1, 2, 3, 4, 5))
print("sum_all(1.5, 2.5)          =", sum_all(1.5, 2.5))

values = [10, 20, 30]
print("sum_all(*values)           =", sum_all(*values), "  # 用 * 解包列表传参")
```

## 实验4：文件读写、异常处理与模块

**思路**：先建立目录与数据文件，再写一个严格的 `parse_student()` 解析函数；
按行处理时用 `try/except` 逐行捕获错误并继续，把有效记录收集成字典列表；
最后做落盘（report.txt、students.pkl）、模块化（student_utils.py）和缺失文件处理。

```python
import os

print("当前工作目录:", os.getcwd())

DATA_DIR = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(DATA_DIR, exist_ok=True)          # 已存在时不报错

DATA_FILE = os.path.join(DATA_DIR, "students.txt")
REPORT_FILE = os.path.join(DATA_DIR, "report.txt")
PKL_FILE = os.path.join(DATA_DIR, "students.pkl")

print("数据目录  :", DATA_DIR)
print("存在      :", os.path.isdir(DATA_DIR))
print("数据文件  :", DATA_FILE)
```

```python
STUDENTS_TXT = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

with open(DATA_FILE, "w", encoding="utf-8") as fh:
    fh.write(STUDENTS_TXT)

with open(DATA_FILE, "r", encoding="utf-8") as fh:
    lines = fh.read().splitlines()

print(f"共 {len(lines)} 行：")
for no, line in enumerate(lines, start=1):
    print(f"{no:>2}| {line}")
```

```python
# 3. parse_student：解析一行，非法记录抛 ValueError
def parse_student(line):
    """解析 'name,score' 形式的一行，返回 {'name': str, 'score': int}。

    非法情况抛 ValueError：
      - 字段数量不为 2
      - 姓名为空
      - 成绩不是整数
      - 成绩不在 0~100 范围内
    """
    fields = line.split(",")

    if len(fields) != 2:
        raise ValueError(f"字段数量应为 2，实际为 {len(fields)}")

    name = fields[0].strip()
    raw_score = fields[1].strip()

    if not name:
        raise ValueError("姓名不能为空")

    try:
        score = int(raw_score)
    except ValueError:
        raise ValueError(f"成绩不是整数：{raw_score!r}") from None

    if not 0 <= score <= 100:
        raise ValueError(f"成绩超出 0~100 范围：{score}")

    return {"name": name, "score": score}


for demo_line in ["Alice,85", "Bob,  58  ", "Grace,abc", "Henry,105", "Incomplete", ",90"]:
    try:
        print(f"{demo_line!r:<16} -> {parse_student(demo_line)}")
    except ValueError as e:
        print(f"{demo_line!r:<16} -> ValueError: {e}")
```

```python
# 4. 按行处理，逐行捕获异常，记录行号与原因，继续处理后续记录
valid_records = []
invalid_records = []

with open(DATA_FILE, "r", encoding="utf-8") as fh:
    next(fh)                                  # 跳过表头
    for lineno, raw_line in enumerate(fh, start=2):
        line = raw_line.strip()
        if not line:                          # 跳过空行
            continue
        try:
            valid_records.append(parse_student(line))
        except ValueError as exc:
            invalid_records.append({
                "line": lineno,
                "content": line,
                "reason": str(exc),
            })

print(f"✅ 有效记录 {len(valid_records)} 条：")
for rec in valid_records:
    print("   ", rec)

print()
print(f"❌ 无效记录 {len(invalid_records)} 条：")
for rec in invalid_records:
    print(f"    第 {rec['line']} 行  {rec['content']!r}  ->  {rec['reason']}")
```

```python
# 5. 统计结果写入 report.txt，并演示 w 与 a 模式
valid_count = len(valid_records)
invalid_count = len(invalid_records)
avg_score = sum(r["score"] for r in valid_records) / valid_count
pass_rate = sum(1 for r in valid_records if r["score"] >= 60) / valid_count

summary_text = (
    f"有效记录数: {valid_count}\n"
    f"无效记录数: {invalid_count}\n"
    f"平均分: {avg_score:.2f}\n"
    f"及格率: {pass_rate:.2%}\n"
)

with open(REPORT_FILE, "w", encoding="utf-8") as fh:      # w：清空后写入
    fh.write(summary_text)

print("--- w 模式（覆盖写入）后 ---")
with open(REPORT_FILE, "r", encoding="utf-8") as fh:
    print(fh.read(), end="")

with open(REPORT_FILE, "a", encoding="utf-8") as fh:      # a：追加，保留原内容
    fh.write("备注: 使用 a 模式追加, 原有内容全部保留\n")

print("--- a 模式（追加写入）后 ---")
with open(REPORT_FILE, "r", encoding="utf-8") as fh:
    print(fh.read(), end="")

print()
print("说明：w 模式会先把文件截断为 0 字节再写入，原有的内容会丢失；")
print("      a 模式从文件末尾继续写，原有内容不受影响。")
```

```python
# 6. pickle 序列化与反序列化
import pickle

with open(PKL_FILE, "wb") as fh:              # 二进制写
    pickle.dump(valid_records, fh)

with open(PKL_FILE, "rb") as fh:              # 二进制读
    restored = pickle.load(fh)

print("恢复的数据:", restored)
print()
print("类型         :", type(restored).__name__)
print("元素个数     :", len(restored), "vs 原", len(valid_records))
print("与原数据相等 :", restored == valid_records)
print("是同一对象吗 :", restored is valid_records, "  # False：反序列化得到的是新对象")
```

## 实验5：类、继承与对象复制

**思路**：`Student` 封装姓名与成绩；`GraduateStudent` 通过 `super().__init__()` 复用父类初始化；
最后用三种方式处理「含嵌套列表的对象」，观察别名与拷贝的差别。

```python
# 1. 定义 Student 类
class Student:
    """学生：姓名 + 成绩。"""

    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        """成绩大于等于 60 分视为及格。"""
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name!r}, score={self.score})"

    def __repr__(self):
        return self.__str__()


demo_student = Student("Alice", 85)
print(demo_student)
print("is_passed():", demo_student.is_passed())
print("不及格示例 :", Student("Bob", 58), "->", Student("Bob", 58).is_passed())
```

```python
# 2. 为实验4的每条有效记录创建 Student 对象
students = [Student(rec["name"], rec["score"]) for rec in valid_records]

print(f"共创建 {len(students)} 个 Student 对象：")
for st in students:
    status = "及格" if st.is_passed() else "不及格"
    print(f"  {st}  ->  {status}")
```

```python
# 3. 继承：GraduateStudent
class GraduateStudent(Student):
    """研究生：在学生的基础上增加研究方向。"""

    def __init__(self, name, score, research_topic):
        super().__init__(name, score)          # 复用父类的初始化
        self.research_topic = research_topic

    def __str__(self):
        return (f"GraduateStudent(name={self.name!r}, score={self.score}, "
                f"research_topic={self.research_topic!r})")

    def __repr__(self):
        return self.__str__()


grad = GraduateStudent("Alice", 85, "机器学习")
print(grad)
print("继承自 Student 的 is_passed():", grad.is_passed())
print("新增属性 research_topic      :", grad.research_topic)
print()
print("isinstance(grad, GraduateStudent):", isinstance(grad, GraduateStudent))
print("isinstance(grad, Student)        :", isinstance(grad, Student), " # 子类也是父类")
print("属性查找顺序 MRO:", [c.__name__ for c in GraduateStudent.__mro__])
```

```python
# 4. 直接赋值 / copy.copy() / copy.deepcopy()
import copy


def make_original():
    """每次都返回一个全新的原始对象，保证三次比较互不影响。"""
    return {"name": "Alice", "scores": [85, 90]}


print("=" * 60)
print("① 直接赋值（alias）")
print("=" * 60)
original = make_original()
assigned = original                       # 只是多了一个名字
assigned["scores"].append(100)
print("原始对象 original :", original)
print("赋值变量 assigned :", assigned)
print("original is assigned        :", original is assigned)
print("@ 改一个另一个也变 —— 完全是同一个对象。")

print()
print("=" * 60)
print("② 浅拷贝 copy.copy()")
print("=" * 60)
original = make_original()
shallow = copy.copy(original)
shallow["scores"].append(100)
print("原始对象 original :", original)
print("浅拷贝   shallow  :", shallow)
print("original is shallow         :", original is shallow, " # 外层是新对象")
print("original['scores'] is shallow['scores']:",
      original["scores"] is shallow["scores"], " # 内层列表仍是同一个")
print("@ 只复制了外层字典，嵌套的列表被共享，改内层会互相影响。")

print()
print("=" * 60)
print("③ 深拷贝 copy.deepcopy()")
print("=" * 60)
original = make_original()
deep = copy.deepcopy(original)
deep["scores"].append(100)
print("原始对象 original :", original)
print("深拷贝   deep     :", deep)
print("original is deep            :", original is deep)
print("original['scores'] is deep['scores']:",
      original["scores"] is deep["scores"], " # 内层列表也是新对象")
print("@ 外层和内层全部递归复制，两者完全独立。")
```

### 实验5 小结

| 方式 | 外层对象 | 嵌套对象 | 改内层是否影响原对象 |
|---|---|---|---|
| `assigned = original` | 同一个 | 同一个 | ✅ 会 |
| `copy.copy(original)` | 新对象 | **同一个** | ✅ 会 |
| `copy.deepcopy(original)` | 新对象 | 新对象 | ❌ 不会 |

因此：只要对象里含有可变容器（列表、字典、集合等），需要真正独立时应当使用 `copy.deepcopy()`。

## 三、实验总结

1. **运算与类型**：`/` 必得 `float`，`//` 向下取整，`%` 取余，`**` 求幂；`int()` 转换是向零截断。
2. **字符串**：不可变，所有「修改」操作实际都返回新对象；`strip/split/join` 是文本处理的核心工具。
3. **函数与控制流**：`for`/`while`、递归/迭代可以得到互相验证的等价结果；
   边界值（0、越界）是测试条件分支的关键。
4. **容器**：`zip`/`enumerate`/推导式让「按位置配对、带序号遍历、条件筛选」非常简洁；
   直接赋值是别名，`copy()` 是浅拷贝，元组不可变。
5. **文件与异常**：`with open(..., encoding="utf-8")` 保证正确编码与自动关闭；
   逐行 `try/except` 可以让「坏数据」不中断整体流程；`w` 覆盖、`a` 追加。
6. **模块化与对象**：把自己的函数写进 `.py` 模块再 `import`，是复用代码的基本手段；
   继承通过 `super().__init__()` 复用父类逻辑；`deepcopy` 才能彻底切断嵌套对象的共享。
